```python
# Creating a DataFrame

import numpy as np
data = [
    {'Name': 'Tom', 'Country': 'Finland', 'City': 'Helsinki'},
    {'Name': 'David', 'Country': 'UK', 'City': 'London'},
    {'Name': 'John', 'Country':'Sweden', 'City':'Stockholm'}
]
df = pd.DataFrame(data)
df
```

```python
# Adding a new column

weights = [74, 78, 69]
df['Weight'] = weights
df
```

```python
# Adding another column
heights = [173, 175, 169]
df['Height'] = heights
df
```

```python
df['Height'] = df['Height'] * 0.01
df
```

```python
df['BMI'] = df['Weight'] / df['Height'] ** 2
df
```

```python
def calculate_bmi ():
    weights = df['Weight']
    heights = df['Height']
    bmi = []
    for w,h in zip(weights, heights):
        b = w/(h**2)
        bmi.append(b)
    return bmi

bmi = calculate_bmi()

df['BMI'] = bmi
df
```
